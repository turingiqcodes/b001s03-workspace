```python
def exercise_1(n):
    for i in range(1,n+1):
        print(i)
    for i in range(1,n+1):
        if i%3==0:
            print(i,"Fizz")
        if i%5==0:
            print(i,"Buzz")
        if i%3==0 and i%5==0:
            print(i,"FizzBuzz")
n=int(input("Enter the number:"))
exercise_1(n)
```

```python
from math import sqrt
def exercise_2(n):
    end=int(sqrt(n))+1
    if n==0 or n==1:
        print("Not Prime")
    for i in range(2,end):
        if i%n==0:
            print("Not Prime")
            return 
    print("Prime")
n=int(input("Enter the Number:"))
exercise_2(n)      
```

```python
def reverse_order(n):
    rev=0
    while n:
        rem=n%10
        rev=rev*10+rem
        n=n//10
    return rev
n=int(input("Enter the number:"))
print(reverse_order(n))
```

```python
def exercise_4(n):
    for i in range(1,11):
        print(n,"*",i,"=",n*i)
n=int(input("Enter the Number"))
exercise_4(n)      
```
