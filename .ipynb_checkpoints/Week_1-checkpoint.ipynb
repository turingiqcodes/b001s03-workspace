```python
def exercise_1(n):
    for i in range(1,n+1):
        print(i)
    for i in range(1,n+1):
        if i%3==0:
            print(i,"Fizz")
        if i%5==0:
            print(i,"Buzz")
        if i%3==0 and i%5==0:
            print(i,"FizzBuzz")
n=int(input("Enter the number:"))
exercise_1(n)
```

```python
from math import sqrt
def exercise_2(n):
    end=int(sqrt(n))+1
    if n==0 or n==1:
        print("Not Prime")
    for i in range(2,end):
        if i%n==0:
            print("Not Prime")
            return 
    print("Prime")
n=int(input("Enter the Number:"))
exercise_2(n)      
```

```python
def reverse_order(n):
    rev=0
    while n:
        rem=n%10
        rev=rev*10+rem
        n=n//10
    return rev
n=int(input("Enter the number:"))
print(reverse_order(n))
```

```python
def exercise_4(n):
    for i in range(1,11):
        print(n,"*",i,"=",n*i)
n=int(input("Enter the Number"))
exercise_4(n)      
```

```python
def count_positive_negative(numbers):
    # Write your code here
    positive_count=0
    negative_count=0
    i=0
    while i<len(numbers):
        if numbers[i]>0:
            positive_count+=1
            print(numbers[i])
        else:
            negative_count+=1
        i+=1
    return positive_count,negative_count


# Test your code with the following test cases:
print(count_positive_negative([10, -5, 20, -15, 30, 25, -8, 12, -2, 40]))

```

```python
def find_largest(numbers):
    # Write your code here
    initial_value=numbers[0]
    for val in numbers:
        if val>initial_value:
            initial_value=val
    return initial_value        


# Test your code with the following test cases:
print(find_largest([45, 12, 98, 23, 67, 34, 89, 10, 55, 76]))
```

```python
def passed_students(students):
    # Write your code here
    result=[]
    for student,val in students.items():
        if val>=50:
            result.append(student)
    return result        



# Test your code with the following test cases:
print(passed_students({'Ravi': 85, 'Priya': 92, 'Kiran': 45, 'Anil': 55, 'Meena': 40}))
```

```python
def performance_summary(students):
    # Write your code here
    result={}
    pass_c=0
    fail_c=0
    max_score=0
    min_score=(float('inf'))
    for val in students.values():
        if val>=50:
            pass_c+=1
        else:
            fail_c+=1
        max_score=max(max_score,val)
        min_score=min(min_score,val)
        result['passed']=pass_c
        result['failed']=fail_c
        result['higest']=max_score
        result['lowest']=min_score
    return result    





# Test your code with the following test cases:
print(performance_summary({'Ravi': 85, 'Priya': 92, 'Kiran': 45, 'Anil': 55, 'Meena': 40, 'John': 70, 'Asha': 88, 'Vijay': 35, 'Arun': 61, 'Neha': 95}))
```
